```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)

n = 1000

regions = [
    "Pune",
    "Mumbai",
    "Delhi",
    "Bangalore",
    "Hyderabad"
]

issue_types = [
    "High Latency",
    "Call Drop",
    "Slow Internet",
    "Network Outage",
    "No Service"
]

priorities = [
    "Low",
    "Medium",
    "High",
    "Critical"
]

tickets = pd.DataFrame({
    "ticket_id": [
        f"TKT{i:05d}"
        for i in range(1, n + 1)
    ],

    "created_at": pd.date_range(
        start="2025-01-01",
        periods=n,
        freq="h"
    ),

    "region": np.random.choice(
        regions, n
    ),

    "tower_id": [
        f"T{np.random.randint(1, 101):03d}"
        for _ in range(n)
    ],

    "issue_type": np.random.choice(
        issue_types, n
    ),

    "priority": np.random.choice(
        priorities, n
    ),

    "status": np.random.choice(
        ["Resolved", "Open", "In Progress"],
        n,
        p=[0.80, 0.10, 0.10]
    )
})

tickets["resolved_at"] = pd.NaT

mask = tickets["status"] == "Resolved"

tickets.loc[mask, "resolved_at"] = (
    tickets.loc[mask, "created_at"]
    + pd.to_timedelta(
        np.random.randint(1, 49, mask.sum()),
        unit="h"
    )
)

tickets.to_csv(
    "D:\\KirtiDA_Prep\\Projects\\telecom-network-analytics\\data\\raw\\tickets.csv",
    index=False
)

tickets.head()
```

```python
# Added ticket-resolution hours
tickets["created_at"] = pd.to_datetime(
    tickets["created_at"]
)

tickets["resolved_at"] = pd.to_datetime(
    tickets["resolved_at"]
)

tickets["resolution_hours"] = (
    tickets["resolved_at"]
    - tickets["created_at"]
).dt.total_seconds() / 3600
```

```python
# average resolution time
avg_resolution = (
    tickets["resolution_hours"].mean()
)

print(avg_resolution)
```
